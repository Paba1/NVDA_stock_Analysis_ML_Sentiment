## 2. Inżynieria Cech (Feature Engineering) i Etykietowanie Danych
Kluczowym etapem budowy modelu predykcyjnego dla instrumentów finansowych jest transformacja surowych danych szeregów czasowych (OHLCV) w zestaw cech (features), które posiadają silną wartość informacyjną, przy jednoczesnym zachowaniu stacjonarności.

2.1. Transformacja cech i eliminacja błędu skali (Scale Bias)
W analizie historycznej akcji NVIDIA (2017–2025) występuje problem ogromnej różnicy w nominalnej cenie akcji. Standardowe wskaźniki techniczne (np. SMA, EMA, BB) przyjmują wartości w USD, co uniemożliwia modelowi XGBoost naukę uniwersalnych wzorców (model uczyłby się konkretnych poziomów cen zamiast dynamiki ruchu). W związku z tym wdrożono następujące usprawnienia:

Wskaźniki relatywne: Zamiast surowych średnich kroczących, obliczono procentowy dystans ceny od średniej (np. Dist_SMA_20).

Normalizacja zmienności: Zastosowano relatywny wskaźnik ATR (ATR_Relative) oraz szerokość wstęgi Bollingera (BB_Width), co pozwala odnieść zmienność do aktualnego poziomu cenowego.

Analiza przepływów i cykli: Dodano zaawansowane wskaźniki kwantowe, takie jak Chaikin Money Flow (CMF) do śledzenia akumulacji kapitału oraz Hilbert Transform, służący do identyfikacji dominującego cyklu rynkowego.

2.2. Zaawansowane etykietowanie: Triple Barrier Method (TBM)
Klasyczne podejście do klasyfikacji (proste przewidywanie wzrostu/spadku o dowolną wartość) jest obarczone dużym szumem rynkowym. W niniejszej pracy zastosowano metodę Triple Barrier Method (Marcos Lopez de Prado, 2018), która jest standardem w profesjonalnym Machine Learningu finansowym:

Bariera Górna (Profit Take): Dynamiczny próg zysku oparty na aktualnej zmienności rynku (mnożnik zmienności).

Bariera Dolna (Stop Loss): Próg ochrony kapitału. Etykieta -1 informuje model o scenariuszu, w którym strata wystąpiła przed osiągnięciem celu.

Bariera Czasowa (Vertical Barrier): Jeśli w ciągu 5 dni handlowych cena nie przebije żadnej z barier, ruch uznawany jest za nieistotny statystycznie (szum, etykieta 0).

2.3. Selekcja cech i redukcja multikolinearności
Zjawisko multikolinearności (silnej korelacji między cechami) negatywnie wpływa na interpretowalność modelu i stabilność wag.

Wdrożono automatyczny filtr korelacji, który usuwa cechy o współczynniku korelacji Pearsona wyższym niż 0.90.

Z zestawu danych usunięto surowe kolumny cenowe, pozostawiając jedynie cechy stacjonarne (procentowe zmiany i oscylatory), co zapewnia większą zdolność generalizacji modelu na przyszłe dane.

```python
# Przygotowanie danych do TA-Lib
close = df['Close'].to_numpy().astype(float).ravel()
high  = df['High'].to_numpy().astype(float).ravel()
low   = df['Low'].to_numpy().astype(float).ravel()
volume = df['Volume'].to_numpy().astype(float).ravel()

# --- 2. FEATURE ENGINEERING (TWOJE + QUANT) ---
# Uwaga: Każdy wskaźnik cenowy zamieniamy na RELATYWNY (odległość %)
df['Returns'] = np.log(df['Close'] / df['Close'].shift(1))
df['RSI'] = talib.RSI(close, timeperiod=14)
df['ADX'] = talib.ADX(high, low, close, timeperiod=14)

# MACD - interesuje nas tylko Histogram (siła pędu), a nie surowe wartości MACD
_, _, df['MACD_hist'] = talib.MACD(close, fastperiod=12, slowperiod=26, signalperiod=9)

# ATR Relative - zmienność w stosunku do ceny
df['ATR_Relative'] = talib.ATR(high, low, close, timeperiod=14) / df['Close']

# Wskaźniki dystansu (Usuwamy surowe SMA/EMA/BB, zostawiamy % dystansu)
df['Dist_SMA_20'] = (df['Close'] - talib.SMA(close, timeperiod=20)) / talib.SMA(close, timeperiod=20)
df['Dist_EMA_50'] = (df['Close'] - talib.EMA(close, timeperiod=50)) / talib.EMA(close, timeperiod=50)

upper, middle, lower = talib.BBANDS(close, timeperiod=20)
df['BB_Width'] = (upper - lower) / middle
df['Percent_BB'] = (df['Close'] - lower) / (upper - lower)

# Wolumen i Impet
df['Relative_Volume'] = df['Volume'] / df['Volume'].rolling(window=20).mean()
df['CMF'] = talib.ADOSC(high, low, close, volume, fastperiod=3, slowperiod=10) # Chaikin Money Flow
df['ROC_10'] = talib.ROC(close, timeperiod=10) # Rate of Change

# Efektywność i Cykle (Quant Tools)
direction = abs(df['Close'] - df['Close'].shift(10))
volatility = abs(df['Close'] - df['Close'].shift(1)).rolling(window=10).sum()
df['Efficiency_Ratio'] = direction / (volatility + 1e-10)
df['HT_Period'] = talib.HT_DCPERIOD(close) # Cykl rynkowy

# Pomocnicze
df['DayOfWeek'] = df.index.dayofweek
df['Gap'] = (df['Open'] - df['Close'].shift(1)) / df['Close'].shift(1)

# --- 3. TRIPLE BARRIER METHOD (PROFESJONALNY TARGET) ---
def get_tbm_labels(price_series, returns_series, days=5, vol_mult=1.5):
    vol = returns_series.rolling(window=20).std()
    labels = []
    for i in range(len(price_series) - days):
        p_now = price_series.iloc[i]
        b_up = p_now * (1 + vol.iloc[i] * vol_mult)
        b_down = p_now * (1 - vol.iloc[i] * vol_mult)
        
        future = price_series.iloc[i+1 : i+days+1]
        
        reach_up = future[future >= b_up].index.min()
        reach_down = future[future <= b_down].index.min()
        
        if pd.isna(reach_up) and pd.isna(reach_down): labels.append(0) # Neutral/Noise
        elif pd.isna(reach_down) or (not pd.isna(reach_up) and reach_up < reach_down): labels.append(1) # Profit
        else: labels.append(-1) # Stop Loss
    labels.extend([0]*days)
    return labels

df['Target'] = get_tbm_labels(df['Close'], df['Returns'])

# --- 4. FILTROWANIE I CZYSZCZENIE ---
# Usuwamy surowe kolumny cenowe (Open, High, Low, Close, Volume), zostawiamy tylko cechy relatywne
features_to_keep = [col for col in df.columns if col not in ['Open', 'High', 'Low', 'Close', 'Volume']]
df_final = df[features_to_keep].copy()

# Usuwamy NaN powstałe przez wskaźniki
df_final.dropna(inplace=True)

# Automatyczny filtr korelacji (usuwa wskaźniki mówiące to samo)
corr_matrix = df_final.drop(columns=['Target']).corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [column for column in upper.columns if any(upper[column] > 0.90)]
df_final.drop(columns=to_drop, inplace=True)

print(f"--- GOTOWE ---")
print(f"Rozmiar danych: {df_final.shape}")
print(f"Usunięte kolumny (korelacja > 0.9): {to_drop}")
print(f"Rozkład klas w Target:\n{df_final['Target'].value_counts(normalize=True)}")
df_final.head()
```


## 3. Inżynieria Cech i Zaawansowane EDA (Metodologia)
W tej sekcji przygotowujemy dane zgodnie z rygorem analizy szeregów czasowych. Główne założenia metodologiczne:

Eliminacja Price-Scale Bias: Wszystkie wskaźniki techniczne bazujące na cenie zostały znormalizowane do postaci relatywnej (procentowy dystans od ceny lub oscylatory). Pozwala to modelowi uczyć się uniwersalnych wzorców, niezależnie od nominalnej wartości akcji (Nvidia w 2017 vs 2024).

Triple Barrier Method (TBM): Zamiast naiwnego przewidywania kierunku "góra/dół", zastosowano metodę trzech barier (Profit Take, Stop Loss, Time Barrier). Etykieta 1 oznacza, że cena osiągnęła zysk przed stratą, -1 oznacza aktywację Stop Loss, a 0 to szum rynkowy.

Filtracja Redundancji: Usunięto cechy o korelacji > 0.90, aby zapobiec multikolinearności, która zaburza interpretację ważności cech w modelach drzewiastych (XGBoost).

Mutual Information (MI): Wykorzystano MI do nieliniowej oceny wartości predykcyjnej cech, co pozwala odróżnić sygnał od szumu przed trenowaniem modelu.

```python

# To mówi nam: "Która cecha faktycznie ma wpływ na Profit/Loss?"
X_eda = df_final.drop(columns=['Target'])
y_eda = df_final['Target']

mi_scores = mutual_info_classif(X_eda, y_eda, discrete_features=False, random_state=42)
mi_series = pd.Series(mi_scores, index=X_eda.columns).sort_values(ascending=True)

plt.figure(figsize=(10, 8))
mi_series.plot(kind='barh', color='teal')
plt.title("Wartość Informacyjna Cech (Mutual Information Score)")
plt.xlabel("Score (Wyższy = Lepszy Sygnał)")
plt.grid(axis='x', alpha=0.3)
plt.show()

# --- 2. TWOJA HEATMAPA (Ale na sterydach TBM) ---
# Sprawdzamy prawdopodobieństwo uderzenia w Profit (Target=1)
# Najpierw musimy przywrócić na chwilę binowanie RSI do df_final
temp_df = df_final.copy()
temp_df['RSI_bin'] = pd.cut(temp_df['RSI'], bins=[0, 30, 50, 70, 100], labels=['<30', '30-50', '50-70', '>70'])

# Liczymy % sukcesów (Target=1) dla kombinacji RSI i np. Efficiency Ratio (czy trend jest czysty)
temp_df['ER_bin'] = pd.qcut(temp_df['Efficiency_Ratio'], q=4, labels=['Zaszumiony', 'Średni', 'Czysty', 'Idealny'])

prob_table = temp_df[temp_df['Target'] == 1].groupby(['ER_bin', 'RSI_bin'], observed=False).size() / \
             temp_df.groupby(['ER_bin', 'RSI_bin'], observed=False).size()

plt.figure(figsize=(10, 5))
sns.heatmap(prob_table.unstack(), annot=True, cmap='RdYlGn', fmt='.2%')
plt.title('Prawdopodobieństwo uderzenia w PROFIT (Target=1)\nw zależności od RSI i Czystości Trendu (ER)')
plt.show()

# --- 3. ANALIZA AUTOKORELACJI (ACF) ---
# Czy dzisiejszy Target zależy od wczorajszego? 
plt.figure(figsize=(10, 4))
plot_acf(df_final['Target'], lags=15)
plt.title("Autokorelacja Targetu (Czy ruchy Nvidii się powtarzają?)")
plt.show()
```

#### 4. Budowa dwóch modeli,jeden podstawowy bez uwzględnienai sekcji EDA drugi zoptymalizowany (usunięcie kolumn na podstawie EDA - dane testowe) 

```python

def run_trading_model(X_data, y_data, model_label="Model"):
    """
    Ujednolicona funkcja do trenowania i ewaluacji modeli.
    Gwarantuje identyczny SEED i podział danych dla każdego wywołania.
    """
    print(f"\n{'='*40}\nPROCESOWANIE: {model_label}\n{'='*40}")
    
    # 1. STAŁA KONFIGURACJA (Gwarancja powtarzalności)
    SEED = 42
    
    # 2. PODZIAŁ CHRONOLOGICZNY (Identyczny dla każdego wywołania)
    # Mapowanie: -1 -> 0, 0 -> 1, 1 -> 2 (wymóg XGBoost dla klasyfikacji wieloklasowej)
    y_mapped = y_data + 1
    
    # Podział 80/20 bez mieszania (chronologiczny)
    split_idx = int(len(X_data) * 0.8)
    X_train, X_test = X_data.iloc[:split_idx], X_data.iloc[split_idx:]
    y_train, y_test = y_mapped.iloc[:split_idx], y_mapped.iloc[split_idx:]
    
    # 3. KONFIGURACJA WALIDACJI CZASOWEJ (TimeSeriesSplit)
    # n_splits=5 symuluje 5 różnych okresów rynkowych do testów wewnętrznych
    tscv = TimeSeriesSplit(n_splits=5)
    
    # 4. MODEL (Bez zbędnych parametrów wywołujących warningi)
    xgb = XGBClassifier(
        objective='multi:softprob',
        random_state=SEED,
        eval_metric='mlogloss'
    )
    
    # Przestrzeń poszukiwań (regularyzacja Gamma i Subsample chronią przed szumem)
    param_grid = {
        'n_estimators': [100, 200, 300],
        'max_depth': [3, 4, 5],
        'learning_rate': [0.01, 0.05, 0.1],
        'gamma': [1, 5],             # Im wyższa, tym model "ostrożniejszy"
        'subsample': [0.8, 0.9],     # Każde drzewo widzi tylko 80-90% danych (zapobiega overfittingowi)
        'colsample_bytree': [0.8, 0.9]
    }
    
    # 5. TRENING Z OPTYMALIZACJĄ (Używamy SEED przy losowaniu parametrów)
    search = RandomizedSearchCV(
        estimator=xgb, 
        param_distributions=param_grid, 
        cv=tscv, 
        n_iter=10, 
        scoring='f1_macro', 
        n_jobs=-1, 
        random_state=SEED
    )
    
    print(f"Optymalizacja modelu {model_label} w toku...")
    search.fit(X_train, y_train)
    best_mod = search.best_estimator_
    
    # 6. EVALUACJA NA DANYCH TESTOWYCH (Out-of-Sample)
    y_pred = best_mod.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    
    print(f"WYNIK ACCURACY [{model_label}]: {acc:.2%}")
    print("\nSZCZEGÓŁOWY RAPORT KLASYFIKACJI:")
    print(classification_report(y_test, y_pred, target_names=['StopLoss', 'Neutral', 'Profit']))
    
    # 7. INTERPRETACJA MODELU (SHAP VALUES)
    # SHAP tłumaczy, jak każda cecha wpłynęła na finalną decyzję modelu
    print(f"\nGenerowanie wykresów SHAP dla {model_label}...")
    explainer = shap.TreeExplainer(best_mod)
    shap_vals = explainer.shap_values(X_test)
    
    # Obsługa formatów SHAP (naprawa AssertionError dla różnych wersji bibliotek)
    if isinstance(shap_vals, list):
        # Starszy format (lista macierzy dla każdej klasy) - bierzemy indeks [2] dla Profit
        shap_profit = shap_vals[2]
    elif len(shap_vals.shape) == 3:
        # Nowszy format (macierz 3D: [próbki, cechy, klasy])
        shap_profit = shap_vals[:, :, 2]
    else:
        # Format awaryjny
        shap_profit = shap_vals

    plt.figure(figsize=(10, 6))
    shap.summary_plot(shap_profit, X_test, plot_type="dot", show=False)
    plt.title(f"Wpływ cech na PROFIT (Beeswarm) - {model_label}")
    plt.show()
    
    # --- ZWRACAMY KOMPLET DANYCH DO BACKTESTINGU I PORÓWNAŃ ---
    return best_mod, y_pred, X_test, y_test

# --- URUCHOMIENIE EKSPERYMENTU PORÓWNAWCZEGO ---

# MODEL 1: Pełny zestaw cech
model_full, pred_full, xtest_full, ytest_full = run_trading_model(
    df_final.drop(columns=['Target']), 
    df_final['Target'], 
    "MODEL PEŁNY (All Features)"
)

# MODEL 2: Selekcja na podstawie MI Score (> 0.005)
threshold = 0.005
top_feats = mi_series[mi_series > threshold].index.tolist()

model_opt, pred_opt, xtest_opt, ytest_opt = run_trading_model(
    df_final[top_feats], 
    df_final['Target'], 
    "MODEL ZOPTYMALIZOWANY (MI Selection)"
)
```

# Symulacja tradingu za pomcą modelu

 Hipoteza Efektywnego Rynku: Nvidia jest tak płynną i analizowaną spółką, że niemal wszystkie informacje techniczne są już "w cenie".

Overfitting na szumie: Zaawansowane metody analizy technicznej często wyłapują "szum", a nie realny trend.

Paradoks Aktywnego Inwestowania: Twoje badanie potwierdza tezę, że w silnym trendzie wzrostowym (Big Tech Bull Market) proste strategie pasywne są ekstremalnie trudne do pobicia przez algorytmy ML operujące na danych dziennych.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def run_backtest_realistic_analytics(y_test_mapped, y_pred, prices_test, initial_capital=10000, fee=0.001):
    """
    Pełna analityka dla modelu realistycznego:
    - Uwzględnia STRATY (brak clip)
    - Uwzględnia PROWIZJE I SPREAD (fee)
    - Pełne statystyki ryzyka
    """
    # 1. Mapowanie predykcji na sygnały: Long (1), Neutral (0), Short (-1)
    # y_pred - 1 zamienia klasy [0, 1, 2] na [-1, 0, 1]
    predictions = pd.Series(y_pred - 1, index=prices_test.index)
    
    # 2. Obliczanie stóp zwrotu akcji
    returns = prices_test.pct_change().fillna(0)
    
    # 3. Wykrywanie transakcji (każda zmiana pozycji kosztuje fee)
    trades = (predictions.diff().fillna(0) != 0).astype(int)
    
    # 4. Obliczanie zwrotów strategii (Sygnał * Zwrot - Koszt transakcji)
    # Tutaj NIE MA .clip(lower=0), więc straty są w pełni widoczne
    strategy_returns = (predictions * returns) - (trades * fee)
    
    # 5. Budowa krzywej kapitału i metryk
    equity_curve = (1 + strategy_returns).cumprod() * initial_capital
    buy_and_hold = (1 + returns).cumprod() * initial_capital
    
    # Metryki ryzyka
    total_return = (equity_curve.iloc[-1] / initial_capital - 1) * 100
    bh_return = (buy_and_hold.iloc[-1] / initial_capital - 1) * 100
    
    # Max Drawdown (największy spadek kapitału)
    rolling_max = equity_curve.cummax()
    drawdown = (equity_curve - rolling_max) / rolling_max
    max_drawdown = drawdown.min() * 100
    
    # Sharpe Ratio (roczny)
    sharpe = np.sqrt(252) * (strategy_returns.mean() / strategy_returns.std()) if strategy_returns.std() != 0 else 0

    # 6. Wizualizacja (Kapitał + Drawdown)
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10), gridspec_kw={'height_ratios': [3, 1]})
    
    ax1.plot(equity_curve.values, label=f'Strategia ML Realistyczna (Fee {fee*100}%)', color='red', linewidth=2)
    ax1.plot(buy_and_hold.values, label='Buy & Hold (Nvidia)', color='gray', linestyle='--', alpha=0.5)
    ax1.set_title('REALISTYCZNA Analiza Wydajności (Uwzględnia koszty i straty)', fontsize=14)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    ax2.fill_between(range(len(drawdown)), drawdown.values * 100, 0, color='red', alpha=0.2)
    ax2.set_title('Realne Obsunięcie Kapitału (%)', fontsize=12)
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

    # 7. Raport końcowy do tabeli w pracy
    print(f"{'='*45}")
    print(f"   RAPORT REALISTYCZNY (Z PROWIZJAMI)")
    print(f"{'='*45}")
    print(f"Wynik Strategii:            {total_return:.2f}%")
    print(f"Wynik Buy & Hold:           {bh_return:.2f}%")
    print(f"Kapitał końcowy:            {equity_curve.iloc[-1]:.2f} USD")
    print(f"Maksymalne obsunięcie (MDD): {max_drawdown:.2f}%")
    print(f"Wskaźnik Sharpe'a:          {sharpe:.2f}")
    print(f"Liczba transakcji:          {trades.sum()}")
    print(f"Suma opłat prowizyjnych:    {(trades.sum() * fee * 100):.2f}% kapitału")
    print(f"{'='*45}")

    return equity_curve

# URUCHOMIENIE
prices_test = df.loc[xtest_opt.index, 'Close']
equity_real = run_backtest_realistic_analytics(ytest_opt, pred_opt, prices_test, fee=0.001)
```

```python
def run_backtest_analytics(best_model, X_test, prices_test, initial_capital=10000, fee=0.001):
    """
    Zaawansowany backtester z dynamicznym sizingiem i pełną analityką.
    """
    # 1. Obliczenia prawdopodobieństwa i sygnałów
    probs = best_model.predict_proba(X_test)
    y_pred = best_model.predict(X_test)
    
    # Kierunek: Profit(2)->1, Neutral(1)->0, StopLoss(0)->-1
    direction = pd.Series(y_pred - 1, index=prices_test.index)
    
    # Pewność dla klasy Profit (pozycja długa)
    conf_profit = probs[:, 2]
    
    # 2. Dynamiczne zaangażowanie (Position Sizing)
    # Wykorzystujemy prawdopodobieństwo jako wagę kapitału
    position_size = np.where(direction != 0, conf_profit, 0)
    position_size = pd.Series(position_size, index=prices_test.index)
    
    # 3. Koszty i Zwroty
    returns = prices_test.pct_change().fillna(0)
    trades = (direction.diff().fillna(0) != 0).astype(int)
    
    # Zwrot netto: (Kierunek * Wielkość * Zmiana) - (Czy był handel * Prowizja)
    strategy_returns = (direction * position_size * returns) - (trades * fee)
    
    # 4. Krzywa kapitału i metryki
    equity_curve = (1 + strategy_returns).cumprod() * initial_capital
    buy_and_hold = (1 + returns).cumprod() * initial_capital
    
    # Metryki ryzyka
    total_return = (equity_curve.iloc[-1] / initial_capital - 1) * 100
    bh_return = (buy_and_hold.iloc[-1] / initial_capital - 1) * 100
    
    # Max Drawdown (maksymalny spadek od szczytu)
    rolling_max = equity_curve.cummax()
    drawdown = (equity_curve - rolling_max) / rolling_max
    max_drawdown = drawdown.min() * 100
    
    # Sharpe Ratio (uproszczony roczny, przy założeniu 252 dni handlowych)
    sharpe = np.sqrt(252) * (strategy_returns.mean() / strategy_returns.std()) if strategy_returns.std() != 0 else 0

    # 5. Wykresy
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10), gridspec_kw={'height_ratios': [3, 1]})
    
    ax1.plot(equity_curve.values, label='Strategia ML (Smart Sizing)', color='#1f77b4', linewidth=2)
    ax1.plot(buy_and_hold.values, label='Buy & Hold', color='#d62728', linestyle='--', alpha=0.5)
    ax1.set_title('Analiza Wydajności Kapitałowej', fontsize=14)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    ax2.fill_between(range(len(drawdown)), drawdown.values * 100, 0, color='red', alpha=0.3)
    ax2.set_title('Obsunięcie Kapitału (Drawdown %)', fontsize=12)
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

    # 6. Wydruk raportu
    print(f"{'='*45}")
    print(f"   RAPORT ANALITYCZNY STRATEGII ML")
    print(f"{'='*45}")
    print(f"Wynik Strategii Smart:      {total_return:.2f}%")
    print(f"Wynik Buy & Hold (Nvidia):  {bh_return:.2f}%")
    print(f"Kapitał końcowy:            {equity_curve.iloc[-1]:.2f} USD")
    print(f"Maksymalne obsunięcie (MDD): {max_drawdown:.2f}%")
    print(f"Wskaźnik Sharpe'a:          {sharpe:.2f}")
    print(f"Średnie zaangażowanie:      {position_size.mean():.2%}")
    print(f"Max pewność modelu:         {conf_profit.max():.2%}")
    print(f"Liczba transakcji:          {trades.sum()}")
    print(f"{'='*45}")

    return equity_curve

# URUCHOMIENIE
prices_test = df.loc[xtest_opt.index, 'Close']
equity_final = run_backtest_analytics(model_opt, xtest_opt, prices_test)
```
